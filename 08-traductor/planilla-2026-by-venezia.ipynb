# TP Traductor: planilla
---

### **Identificación**: (nombre-año-by-apellido. Ej: nossa-2026-by-pandolfo)

`planilla-2026-by-venezia`

### **Objetivo**: (oración que empieza con verbo)

Traducir formularios descritos con palabras simples en español, organizados por sangría, a código HTML con estilos listo para usar en una página web.

### **Alcance**: (viñetas, describiendo funcionalidades)

**El traductor hace:**

* Un formulario por programa, con título, método de envío (`post` o `get`) y ruta de destino.
* **Estructura por sangría obligatoria** (2 espacios por nivel, como en Python): formulario → secciones → filas → campos.
* **Secciones** con título que agrupan campos.
* **Filas**: dos o más campos ubicados uno al lado del otro.
* Tipos de campo: `texto`, `email`, `numero`, `fecha`, `area` (texto largo), `clave`, `opciones` (lista desplegable) y `casilla`.
* Atributos de campo: `nombre` (etiqueta), `placeholder`, `ayuda` (texto chico debajo del campo), `obligatorio`, `min`, `max`, `mensaje` (mensaje de error propio) y lista de valores para `opciones`.
* **Tema** predefinido para todo el formulario: `claro` (por defecto), `oscuro` o `azul`.
* Estilo **card** con espaciado mínimo.
* Comentarios de línea con `#`.
* La traducción a HTML/CSS la realiza un LLM local (ollama) a partir de los datos que arma el parser. El HTML final se escribe **formateado con sangría**.

**El traductor NO hace:**

* Envío real de datos ni conexión con un servidor: solo genera el formulario.
* Campos condicionales, validaciones entre campos ni otra lógica en JavaScript.
* Formularios de varios pasos o varias páginas.
* Estilos personalizados fuera de los temas predefinidos.
* Subida de archivos.
* Variables, condicionales, bucles ni funciones: es un lenguaje de dominio específico, no un lenguaje de programación general.

### **Especificaciones sintácticas**: (GIC en BNF)

```
<Programa>    ::= <Encabezado> INDENTA <Cuerpo> <Boton> DESINDENTA
<Encabezado>  ::= FORMULARIO CADENA <Metodo> IGUAL CADENA <Tema> NUEVA_LINEA
<Metodo>      ::= POST
<Metodo>      ::= GET
<Tema>        ::= TEMA IGUAL <Color>
<Tema>        ::= λ
<Color>       ::= CLARO
<Color>       ::= OSCURO
<Color>       ::= AZUL
<Cuerpo>      ::= <Cuerpo> <Bloque>
<Cuerpo>      ::= <Bloque>
<Bloque>      ::= <Seccion>
<Bloque>      ::= <Elemento>
<Seccion>     ::= SECCION CADENA NUEVA_LINEA INDENTA <Elementos> DESINDENTA
<Elementos>   ::= <Elementos> <Elemento>
<Elementos>   ::= <Elemento>
<Elemento>    ::= <Campo>
<Elemento>    ::= <Fila>
<Fila>        ::= FILA NUEVA_LINEA INDENTA <Campos> DESINDENTA
<Campos>      ::= <Campos> <Campo>
<Campos>      ::= <Campo>
<Campo>       ::= <Tipo> <Atributos> NUEVA_LINEA
<Tipo>        ::= TEXTO | EMAIL | NUMERO | FECHA | AREA | CLAVE | OPCIONES | CASILLA
<Atributos>   ::= <Atributos> <Atributo>
<Atributos>   ::= λ
<Atributo>    ::= NOMBRE IGUAL CADENA
<Atributo>    ::= PLACEHOLDER IGUAL CADENA
<Atributo>    ::= AYUDA IGUAL CADENA
<Atributo>    ::= MENSAJE IGUAL CADENA
<Atributo>    ::= MIN IGUAL ENTERO
<Atributo>    ::= MAX IGUAL ENTERO
<Atributo>    ::= OBLIGATORIO
<Atributo>    ::= PAR_A <Valores> PAR_C
<Valores>     ::= <Valores> COMA CADENA
<Valores>     ::= CADENA
<Boton>       ::= BOTON CADENA NUEVA_LINEA
```

* Símbolos no terminales: `<Programa>`, `<Encabezado>`, `<Metodo>`, `<Tema>`, `<Color>`, `<Cuerpo>`, `<Bloque>`, `<Seccion>`, `<Elementos>`, `<Elemento>`, `<Fila>`, `<Campos>`, `<Campo>`, `<Tipo>`, `<Atributos>`, `<Atributo>`, `<Valores>`, `<Boton>`
* Símbolos terminales: los tokens de la grilla léxica
* Axioma: `<Programa>`
* λ representa la cadena vacía: el tema es opcional y un campo puede no tener atributos (aunque `nombre` se exige en la semántica).
* Producciones recursivas: `<Cuerpo>`, `<Elementos>`, `<Campos>`, `<Atributos>` y `<Valores>` → el lenguaje admite infinitos programas fuente.
* La gramática impide que una sección contenga otra sección y que una fila contenga secciones u otras filas.

### **Especificaciones semánticas**: (grilla explicando acciones semánticas por cada regla de la GIC)

| Regla | Acciones semánticas |
| -- | -- |
| `<Programa> ::= <Encabezado> INDENTA <Cuerpo> <Boton> DESINDENTA` | Arma el documento HTML completo: el `<form>` con el encabezado, el cuerpo y el botón, más el `<style>` con la estructura fija del estilo card (espaciado, anchos, filas) y los colores del tema. Formatea el HTML y el CSS con sangría y lo escribe en el archivo de salida. |
| `<Encabezado> ::= FORMULARIO CADENA <Metodo> IGUAL CADENA <Tema> NUEVA_LINEA` | Genera la apertura `<form method="..." action="...">` con la ruta de la segunda cadena y el título (primera cadena) como encabezado `<h2>`. |
| `<Metodo> ::= POST` / `GET` | Indica el valor del atributo `method` del formulario. |
| `<Tema> ::= TEMA IGUAL <Color>` | Indica qué paleta de colores aplicar a la card. |
| `<Tema> ::= λ` | Se usa el tema `claro`. |
| `<Color> ::= CLARO` / `OSCURO` / `AZUL` | El LLM genera los colores a partir de la descripción del tema. Claro: página gris muy claro, tarjeta blanca, texto gris oscuro. Oscuro: página casi negra, tarjeta gris oscuro, texto claro. Azul: página celeste muy claro, tarjeta blanca, detalles y botón en azul. |
| `<Cuerpo> ::= <Cuerpo> <Bloque>` / `<Bloque>` | Concatena los bloques en el orden en que fueron escritos. |
| `<Bloque> ::= <Seccion>` / `<Elemento>` | Agrega la sección o el elemento al cuerpo del formulario. |
| `<Seccion> ::= SECCION CADENA NUEVA_LINEA INDENTA <Elementos> DESINDENTA` | Genera un `<fieldset>` con la cadena como `<legend>` y adentro sus elementos. |
| `<Elementos> ::= <Elementos> <Elemento>` / `<Elemento>` | Concatena los elementos de la sección en orden. |
| `<Elemento> ::= <Campo>` / `<Fila>` | Agrega el campo o la fila. |
| `<Fila> ::= FILA NUEVA_LINEA INDENTA <Campos> DESINDENTA` | Genera un contenedor que ubica sus campos uno al lado del otro, repartiendo el ancho en partes iguales. **Error semántico** si la fila tiene menos de 2 campos. |
| `<Campos> ::= <Campos> <Campo>` / `<Campo>` | Concatena los campos de la fila en orden. |
| `<Campo> ::= <Tipo> <Atributos> NUEVA_LINEA` | Genera el `<label>` con el `nombre` y el control según el tipo, con sus atributos. El `id`/`name` del control se deriva del `nombre`. **Errores semánticos:** falta `nombre`; `nombre` repetido en el formulario; atributo repetido en el mismo campo; atributo que no corresponde al tipo (ver abajo). |
| `<Tipo> ::= TEXTO` | `<input type="text">` |
| `<Tipo> ::= EMAIL` | `<input type="email">` (el navegador valida el formato de correo) |
| `<Tipo> ::= NUMERO` | `<input type="number">` |
| `<Tipo> ::= FECHA` | `<input type="date">` |
| `<Tipo> ::= AREA` | `<textarea>` |
| `<Tipo> ::= CLAVE` | `<input type="password">` |
| `<Tipo> ::= OPCIONES` | `<select>` con una `<option>` por cada valor de la lista. **Error semántico** si no tiene lista de valores. |
| `<Tipo> ::= CASILLA` | `<input type="checkbox">` con el `nombre` como texto al lado. |
| `<Atributos> ::= <Atributos> <Atributo>` / `λ` | Junta los atributos del campo. |
| `<Atributo> ::= NOMBRE IGUAL CADENA` | Texto de la etiqueta del campo. |
| `<Atributo> ::= PLACEHOLDER IGUAL CADENA` | Texto gris de ejemplo dentro del campo. **Error semántico** en `casilla`, `opciones` y `fecha`. |
| `<Atributo> ::= AYUDA IGUAL CADENA` | Texto chico y gris debajo del campo. |
| `<Atributo> ::= MENSAJE IGUAL CADENA` | Reemplaza el mensaje de error del navegador cuando el campo no es válido. |
| `<Atributo> ::= MIN IGUAL ENTERO` / `MAX IGUAL ENTERO` | Valor mínimo / máximo permitido. **Error semántico** si el campo no es `numero` o si `min` > `max`. |
| `<Atributo> ::= OBLIGATORIO` | El campo no puede quedar vacío (en `casilla`: debe estar marcada). |
| `<Atributo> ::= PAR_A <Valores> PAR_C` | Lista de valores de `opciones`. **Error semántico** si el campo no es `opciones`. |
| `<Valores> ::= <Valores> COMA CADENA` / `CADENA` | Junta los valores en el orden escrito. |
| `<Boton> ::= BOTON CADENA NUEVA_LINEA` | Genera `<button type="submit">` con la cadena como texto. |

Las verificaciones marcadas como **error semántico** las realiza el traductor antes de generar el HTML; si hay alguna, no se genera el formulario y se informan todos los errores con su línea. La generación del HTML de cada regla y de los colores del tema la realiza el LLM (ollama): a cada pedido se le describe en lenguaje natural el significado de la construcción y el elemento que le corresponde según esta grilla; la respuesta se verifica y, si no cumple, se le pide al LLM que la corrija (hasta 3 intentos).

### **Ejemplo simple**

**Programa de entrada** (`01_minimo.planilla`):
```

### **Implementación**: (código)

| Elección | Herramienta |
| -- | -- |
| Lenguaje base | Python 3 |
| Metacompilador | `ply.lex` + `ply.yacc` (parser LALR) |
| LLM | ollama con el modelo `llama3.2` (local, sin internet) |
| Lectura y verificación de HTML | BeautifulSoup |
| Entorno | Jupyter Notebook |

**Cómo funciona el traductor:**

1. **Análisis léxico** (`lexer_planilla.py`): PLY corta el programa en tokens. Además, un filtro con una **pila de niveles** compara la sangría de cada línea con la anterior y genera los tokens `INDENTA` / `DESINDENTA` (como llaves que abren y cierran bloques).
2. **Análisis sintáctico y semántico** (`parser_planilla.py`): `ply.yacc` reconoce la GIC (sin conflictos) y arma un **árbol** con los datos del formulario. Durante el análisis se hacen los controles semánticos de la grilla. Si hay errores, se informan todos con su línea y **no se llama al LLM**.
3. **Generación** (`agente_planilla.py`): se recorre el árbol y, por cada construcción, se le pide al LLM su HTML, describiéndole en lenguaje natural el significado y el elemento que le corresponde según la grilla semántica. Cada respuesta se **limpia y se verifica**; si no cumple, se le devuelve al LLM la lista de problemas para que la corrija (hasta 3 intentos). Se usa `temperature=0` para que la misma entrada dé siempre la misma salida. El LLM también genera los colores del tema.
4. **Armado y formateo** (`traductor_planilla.py`): se unen las partes, se agrega la estructura fija del estilo card y se escribe el HTML y el CSS **con sangría, un elemento por línea**.

| Archivo | Qué hace |
| -- | -- |
| `planilla/lexer_planilla.py` | Análisis léxico: tokens, palabras reservadas, sangría y errores léxicos |
| `planilla/parser_planilla.py` | Análisis sintáctico (GIC) + controles semánticos; arma el árbol |
| `planilla/agente_planilla.py` | Pedidos al LLM, limpieza y verificación de las respuestas |
| `planilla/traductor_planilla.py` | Recorre el árbol, arma el documento y lo formatea |
| `planilla/main.py` | Versión de consola: lee `archivo.planilla` y escribe `archivo.planilla.html` |
| `planilla/test_planilla.py` | TestSuite (pytest) |

````python
%%file planilla/lexer_planilla.py
"""Analizador léxico de planilla (PLY).

Además de los tokens comunes, genera tres tokens que marcan la estructura por sangría:
  NUEVA_LINEA  al terminar cada línea con contenido
  INDENTA      cuando una línea tiene 2 espacios más que la anterior (se abre un nivel)
  DESINDENTA   por cada nivel que se cierra cuando la sangría vuelve atrás
"""
import ply.lex as lex

# ---------------------------------------------------------------- palabras reservadas
reservadas = {
    'formulario': 'FORMULARIO', 'seccion': 'SECCION', 'fila': 'FILA', 'boton': 'BOTON',
    'post': 'POST', 'get': 'GET',
    'tema': 'TEMA', 'claro': 'CLARO', 'oscuro': 'OSCURO', 'azul': 'AZUL',
    'texto': 'TEXTO', 'email': 'EMAIL', 'numero': 'NUMERO', 'fecha': 'FECHA',
    'area': 'AREA', 'clave': 'CLAVE', 'opciones': 'OPCIONES', 'casilla': 'CASILLA',
    'nombre': 'NOMBRE', 'placeholder': 'PLACEHOLDER', 'ayuda': 'AYUDA', 'mensaje': 'MENSAJE',
    'min': 'MIN', 'max': 'MAX', 'obligatorio': 'OBLIGATORIO',
}

tokens = [
    'CADENA', 'ENTERO', 'IGUAL', 'PAR_A', 'PAR_C', 'COMA',
    'NUEVA_LINEA', 'INDENTA', 'DESINDENTA',
] + list(reservadas.values())

ESPACIOS_POR_NIVEL = 2

# ---------------------------------------------------------------- reglas simples
t_IGUAL = r'='
t_PAR_A = r'\('
t_PAR_C = r'\)'
t_COMA = r','

# Dentro de una línea solo se ignoran espacios (los tabs son error).
t_ignore = ' '


# ---------------------------------------------------------------- reglas con acciones
# PLY prueba las funciones en el orden en que están escritas.

def t_SALTO(t):
    r'(\n[ ]*(\#[^\n]*)?)+'
    # Un salto de línea, junto con las líneas vacías o de solo comentario que lo siguen.
    # El valor que nos interesa es la sangría de la próxima línea con contenido:
    # la cantidad de espacios después del último '\n'.
    t.lexer.lineno += t.value.count('\n')
    ultima = t.value[t.value.rfind('\n') + 1:]
    t.value = len(ultima) - len(ultima.lstrip(' '))
    return t


def t_COMENTARIO(t):
    r'\#[^\n]*'
    pass  # se descarta


def t_CADENA(t):
    r"'[^'\n]*'"
    t.value = t.value[1:-1]  # sin las comillas
    return t


def t_CADENA_SIN_CERRAR(t):
    r"'[^'\n]*"
    t.lexer.errores.append((t.lineno, columna(t), f"Cadena sin cerrar: {t.value}"))


def t_ENTERO(t):
    r'[0-9]+'
    t.value = int(t.value)
    return t


def t_PALABRA(t):
    r'[A-Za-zÁÉÍÓÚáéíóúÑñÜü_][A-Za-zÁÉÍÓÚáéíóúÑñÜü_0-9]*'
    if t.value in reservadas:
        t.type = reservadas[t.value]
        return t
    t.lexer.errores.append((t.lineno, columna(t), f"Palabra desconocida: '{t.value}'"))


def t_error(t):
    c = t.value[0]
    if c == '\t':
        msg = "No se permiten tabulaciones: usá 2 espacios por nivel"
    else:
        msg = f"Carácter no reconocido: '{c}'"
    t.lexer.errores.append((t.lineno, columna(t), msg))
    t.lexer.skip(1)


def columna(t):
    inicio_linea = t.lexer.lexdata.rfind('\n', 0, t.lexpos) + 1
    return t.lexpos - inicio_linea + 1


# ---------------------------------------------------------------- sangría
def _nuevo(tipo, valor, linea):
    t = lex.LexToken()
    t.type, t.value, t.lineno, t.lexpos = tipo, valor, linea, -1
    return t


def _con_sangria(base):
    """Recorre los tokens de PLY y reemplaza cada SALTO por NUEVA_LINEA + INDENTA/DESINDENTA."""
    niveles = [0]                 # pila con la sangría de cada nivel abierto
    ultimo = 'NUEVA_LINEA'        # al empezar, "venimos" de una línea nueva
    while True:
        t = base.token()
        if t is None:
            break
        if t.type != 'SALTO':
            yield t
            ultimo = t.type
            continue

        sangria, linea = t.value, base.lineno
        if ultimo != 'NUEVA_LINEA':                 # cierra la línea anterior
            yield _nuevo('NUEVA_LINEA', None, linea - 1)
            ultimo = 'NUEVA_LINEA'
        if base.lexpos >= len(base.lexdata):
            break                                   # salto final del archivo: no hay más líneas

        actual = niveles[-1]
        if sangria % ESPACIOS_POR_NIVEL != 0:
            base.errores.append((linea, 1, f"Sangría de {sangria} espacios: debe ser múltiplo de {ESPACIOS_POR_NIVEL}"))
        elif sangria > actual:
            if sangria != actual + ESPACIOS_POR_NIVEL:
                base.errores.append((linea, 1, "La sangría sube más de un nivel de una línea a la siguiente"))
            niveles.append(sangria)
            yield _nuevo('INDENTA', None, linea)
        else:
            while sangria < niveles[-1]:
                niveles.pop()
                yield _nuevo('DESINDENTA', None, linea)
            if sangria != niveles[-1]:
                base.errores.append((linea, 1, "La sangría no coincide con ningún nivel anterior"))

    if ultimo != 'NUEVA_LINEA':
        yield _nuevo('NUEVA_LINEA', None, base.lineno)
    while len(niveles) > 1:                         # cierra los niveles abiertos al final
        niveles.pop()
        yield _nuevo('DESINDENTA', None, base.lineno)


class LexerPlanilla:
    """Lexer con la misma interfaz que usa ply.yacc (input/token), más la lista de errores."""

    def __init__(self):
        self.base = lex.lex()

    def input(self, texto):
        # Se agrega un '\n' adelante para que la primera línea también pase por la regla de sangría.
        self.base.input('\n' + texto)
        self.base.lineno = 0
        self.base.errores = []
        self._gen = _con_sangria(self.base)

    @property
    def errores(self):
        return self.base.errores

    def token(self):
        return next(self._gen, None)

    def __iter__(self):
        return iter(self.token, None)


tokens = tokens + ['SALTO']  # token interno: nunca llega al parser
lexer = LexerPlanilla()
````

````python
%%file planilla/parser_planilla.py
"""Analizador sintáctico de planilla (ply.yacc) + controles semánticos.

El parser no genera HTML: arma un árbol con los datos del formulario (diccionarios de Python)
y verifica las reglas semánticas. La generación del HTML la hace después el agente (LLM),
recorriendo ese árbol, solo si no hubo ningún error.
"""
import ply.yacc as yacc
from lexer_planilla import LexerPlanilla, tokens as _tokens_lexer

# SALTO es interno del lexer: el parser nunca lo recibe.
tokens = [t for t in _tokens_lexer if t != 'SALTO']

TIPOS_CON_MIN_MAX = {'numero'}
TIPOS_SIN_PLACEHOLDER = {'casilla', 'opciones', 'fecha'}


class ErrorPlanilla(Exception):
    """Error de traducción. tipo: 'léxico', 'sintáctico' o 'semántico'. errores: lista de (línea, mensaje)."""

    def __init__(self, tipo, errores):
        self.tipo, self.errores = tipo, errores
        detalle = '\n'.join(f'  línea {l}: {m}' for l, m in errores)
        super().__init__(f'Error {tipo}:\n{detalle}')


_errores_sintacticos = []
_errores_semanticos = []


def error_semantico(linea, mensaje):
    _errores_semanticos.append((linea, mensaje))


# ---------------------------------------------------------------- programa y encabezado

def p_programa(p):
    'programa : encabezado INDENTA cuerpo boton DESINDENTA'
    formulario = p[1]
    formulario['cuerpo'] = p[3]
    formulario['boton'] = p[4]
    # Los nombres de campo no pueden repetirse en todo el formulario.
    vistos = {}
    for campo in campos_de(formulario['cuerpo']):
        nombre = campo['atributos'].get('nombre')
        if nombre is None:
            continue
        if nombre in vistos:
            error_semantico(campo['linea'], f"El nombre '{nombre}' ya se usó en la línea {vistos[nombre]}")
        else:
            vistos[nombre] = campo['linea']
    p[0] = formulario


def p_encabezado(p):
    'encabezado : FORMULARIO CADENA metodo IGUAL CADENA tema NUEVA_LINEA'
    p[0] = {'titulo': p[2], 'metodo': p[3], 'ruta': p[5], 'tema': p[6]}


def p_metodo(p):
    '''metodo : POST
              | GET'''
    p[0] = p[1]


def p_tema(p):
    'tema : TEMA IGUAL color'
    p[0] = p[3]


def p_tema_vacio(p):
    'tema : '
    p[0] = 'claro'  # tema por defecto


def p_color(p):
    '''color : CLARO
             | OSCURO
             | AZUL'''
    p[0] = p[1]


# ---------------------------------------------------------------- cuerpo: secciones, filas y campos

def p_cuerpo(p):
    '''cuerpo : cuerpo bloque
              | bloque'''
    p[0] = p[1] + [p[2]] if len(p) == 3 else [p[1]]


def p_bloque(p):
    '''bloque : seccion
              | elemento'''
    p[0] = p[1]


def p_seccion(p):
    'seccion : SECCION CADENA NUEVA_LINEA INDENTA elementos DESINDENTA'
    p[0] = {'tipo': 'seccion', 'titulo': p[2], 'elementos': p[5], 'linea': p.lineno(1)}


def p_elementos(p):
    '''elementos : elementos elemento
                 | elemento'''
    p[0] = p[1] + [p[2]] if len(p) == 3 else [p[1]]


def p_elemento(p):
    '''elemento : campo
                | fila'''
    p[0] = p[1]


def p_fila(p):
    'fila : FILA NUEVA_LINEA INDENTA campos DESINDENTA'
    if len(p[4]) < 2:
        error_semantico(p.lineno(1), 'Una fila debe tener al menos 2 campos')
    p[0] = {'tipo': 'fila', 'campos': p[4], 'linea': p.lineno(1)}


def p_campos(p):
    '''campos : campos campo
              | campo'''
    p[0] = p[1] + [p[2]] if len(p) == 3 else [p[1]]


def p_campo(p):
    'campo : tipo atributos NUEVA_LINEA'
    control, linea = p[1]
    atributos = {}
    for clave, valor in p[2]:
        if clave in atributos:
            error_semantico(linea, f"El atributo '{clave}' está repetido")
        atributos[clave] = valor
    verificar_campo(control, atributos, linea)
    p[0] = {'tipo': 'campo', 'control': control, 'atributos': atributos, 'linea': linea}


def p_tipo(p):
    '''tipo : TEXTO
            | EMAIL
            | NUMERO
            | FECHA
            | AREA
            | CLAVE
            | OPCIONES
            | CASILLA'''
    p[0] = (p[1], p.lineno(1))


def p_atributos(p):
    'atributos : atributos atributo'
    p[0] = p[1] + [p[2]]


def p_atributos_vacio(p):
    'atributos : '
    p[0] = []


def p_atributo_cadena(p):
    '''atributo : NOMBRE IGUAL CADENA
                | PLACEHOLDER IGUAL CADENA
                | AYUDA IGUAL CADENA
                | MENSAJE IGUAL CADENA'''
    p[0] = (p[1], p[3])


def p_atributo_entero(p):
    '''atributo : MIN IGUAL ENTERO
                | MAX IGUAL ENTERO'''
    p[0] = (p[1], p[3])


def p_atributo_obligatorio(p):
    'atributo : OBLIGATORIO'
    p[0] = ('obligatorio', True)


def p_atributo_valores(p):
    'atributo : PAR_A valores PAR_C'
    p[0] = ('valores', p[2])


def p_valores(p):
    '''valores : valores COMA CADENA
               | CADENA'''
    p[0] = p[1] + [p[3]] if len(p) == 4 else [p[1]]


def p_boton(p):
    'boton : BOTON CADENA NUEVA_LINEA'
    p[0] = p[2]


# ---------------------------------------------------------------- errores sintácticos

NOMBRES_TOKENS = {
    'NUEVA_LINEA': 'fin de línea', 'INDENTA': 'aumento de sangría',
    'DESINDENTA': 'fin de bloque (menos sangría)', 'IGUAL': "'='",
    'PAR_A': "'('", 'PAR_C': "')'", 'COMA': "','",
}


def describir(tok):
    if tok.type == 'CADENA':
        return f"el texto '{tok.value}'"
    if tok.type == 'ENTERO':
        return f'el número {tok.value}'
    return NOMBRES_TOKENS.get(tok.type, f"la palabra '{tok.value}'")


def p_error(tok):
    if tok is None:
        _errores_sintacticos.append((None, 'El programa terminó antes de tiempo (¿falta el botón?)'))
    else:
        _errores_sintacticos.append((tok.lineno, f'No se esperaba {describir(tok)}'))
    raise _CortarAnalisis()


class _CortarAnalisis(Exception):
    pass


# ---------------------------------------------------------------- controles semánticos

def verificar_campo(control, atributos, linea):
    if 'nombre' not in atributos:
        error_semantico(linea, f"El campo '{control}' no tiene nombre")
    if ('min' in atributos or 'max' in atributos) and control not in TIPOS_CON_MIN_MAX:
        error_semantico(linea, f"'min' y 'max' solo valen en campos 'numero', no en '{control}'")
    if 'min' in atributos and 'max' in atributos and atributos['min'] > atributos['max']:
        error_semantico(linea, f"'min' ({atributos['min']}) es mayor que 'max' ({atributos['max']})")
    if control == 'opciones' and 'valores' not in atributos:
        error_semantico(linea, "Un campo 'opciones' necesita su lista de valores, por ejemplo ('A', 'B')")
    if control != 'opciones' and 'valores' in atributos:
        error_semantico(linea, f"La lista de valores solo vale en campos 'opciones', no en '{control}'")
    if 'placeholder' in atributos and control in TIPOS_SIN_PLACEHOLDER:
        error_semantico(linea, f"'placeholder' no vale en campos '{control}'")


def campos_de(bloques):
    """Devuelve todos los campos del formulario, estén sueltos, en secciones o en filas."""
    for b in bloques:
        if b['tipo'] == 'campo':
            yield b
        elif b['tipo'] == 'fila':
            yield from b['campos']
        else:
            yield from campos_de(b['elementos'])


# ---------------------------------------------------------------- punto de entrada

parser = yacc.yacc(debug=False, write_tables=False)
_lexer = LexerPlanilla()


def analizar(texto):
    """Analiza un programa planilla. Devuelve el árbol del formulario o lanza ErrorPlanilla."""
    _errores_sintacticos.clear()
    _errores_semanticos.clear()
    arbol = None
    try:
        arbol = parser.parse(texto, lexer=_lexer)
    except _CortarAnalisis:
        pass
    # Se informa primero el error de la etapa más temprana.
    if _lexer.errores:
        raise ErrorPlanilla('léxico', [(l, f'{m} (columna {c})') for l, c, m in _lexer.errores])
    if _errores_sintacticos:
        raise ErrorPlanilla('sintáctico', list(_errores_sintacticos))
    if _errores_semanticos:
        raise ErrorPlanilla('semántico', sorted(_errores_semanticos, key=lambda e: e[0]))
    return arbol
````

````python
%%file planilla/agente_planilla.py
"""Agente de planilla: le pide al LLM (ollama) el HTML y el CSS de cada parte del formulario.

Cada pedido describe en lenguaje natural el significado de la construcción (la especificación
semántica) y el elemento HTML que le corresponde según la grilla semántica. La respuesta se limpia
y se verifica; si tiene problemas, se le devuelve al LLM la lista de problemas para que la corrija.
"""
import re
import ollama
from bs4 import BeautifulSoup
from bs4.formatter import HTMLFormatter

MODELO = 'llama3.2'
OPCIONES = {'temperature': 0, 'seed': 7}   # sin azar: misma entrada, misma salida
INTENTOS = 3
MARCA = '{{CONTENIDO}}'                    # lugar donde después se insertan los elementos hijos

SISTEMA = ("Sos un generador de formularios web. Respondé ÚNICAMENTE con el código pedido: "
           "sin explicaciones, sin markdown, sin ``` y sin etiquetas <html>, <head> ni <body>. "
           "Usá exactamente los elementos, id, name, class y atributos indicados, sin agregar otros.")


class ErrorAgente(Exception):
    pass


# ---------------------------------------------------------------- comunicación con el LLM

def limpiar(respuesta, es_css=False):
    """Saca lo que el LLM agrega de más: bloques ``` y texto antes o después del código."""
    r = re.sub(r'```[a-zA-Z]*', '', respuesta).strip()
    if es_css:
        return re.sub(r'</?style[^>]*>', '', r).strip()
    inicio = r.find('<')
    if inicio == -1:
        return r
    fin = max(r.rfind('>') + 1, r.rfind(MARCA) + len(MARCA) if MARCA in r else 0)
    # BeautifulSoup cierra las etiquetas que el LLM haya dejado abiertas (por ejemplo </form>).
    return BeautifulSoup(r[inicio:fin], 'html.parser').decode(formatter=_SIN_ORDENAR)


class _FormatoSinOrdenar(HTMLFormatter):
    def attributes(self, tag):          # respeta el orden de los atributos que escribió el LLM
        return list(tag.attrs.items())


_SIN_ORDENAR = _FormatoSinOrdenar()


def atributos_de_mas(elemento, permitidos):
    sobran = sorted(set(elemento.attrs) - set(permitidos))
    return [f'el <{elemento.name}> tiene atributos que no se pidieron: {", ".join(sobran)}'] if sobran else []


_memoria = {}  # pedido -> respuesta ya verificada (evita repetir pedidos idénticos)


def pedir(pedido, verificar, es_css=False):
    """Hace el pedido, verifica la respuesta y, si hay problemas, pide la corrección."""
    if pedido not in _memoria:
        _memoria[pedido] = _pedir_al_llm(pedido, verificar, es_css)
    return _memoria[pedido]


def _pedir_al_llm(pedido, verificar, es_css):
    mensajes = [{'role': 'system', 'content': SISTEMA}, {'role': 'user', 'content': pedido}]
    problemas = []
    for _ in range(INTENTOS):
        respuesta = ollama.chat(model=MODELO, messages=mensajes, options=OPCIONES)['message']['content']
        codigo = limpiar(respuesta, es_css)
        problemas = verificar(codigo)
        if not problemas:
            return codigo
        mensajes += [
            {'role': 'assistant', 'content': respuesta},
            {'role': 'user', 'content': 'Tu respuesta tiene estos problemas:\n- ' + '\n- '.join(problemas)
                                        + '\nCorregilos y devolvé de nuevo el código COMPLETO (también lo que ya estaba bien), sin explicaciones.'},
        ]
    raise ErrorAgente(f'El LLM no generó código válido después de {INTENTOS} intentos: {"; ".join(problemas)}')


# ---------------------------------------------------------------- campos

# tipo de campo -> (significado en lenguaje natural, elemento HTML según la grilla semántica)
CONTROLES = {
    'texto':    ('un campo de texto de una sola línea', '<input type="text">'),
    'email':    ('un campo para escribir un correo electrónico (el navegador valida el formato)', '<input type="email">'),
    'numero':   ('un campo numérico', '<input type="number">'),
    'fecha':    ('un campo para elegir una fecha', '<input type="date">'),
    'area':     ('un campo de texto largo, de varias líneas', '<textarea>'),
    'clave':    ('un campo de contraseña (los caracteres se ocultan al escribir)', '<input type="password">'),
    'opciones': ('una lista desplegable para elegir una opción', '<select>'),
    'casilla':  ('una casilla de verificación', '<input type="checkbox">'),
}


def pedido_campo(campo, id_):
    control, at = campo['control'], campo['atributos']
    significado, elemento = CONTROLES[control]
    lineas = [f'Generá un campo de formulario envuelto en un <div class="campo">.',
              f'Es {significado}: usá el elemento {elemento} con id="{id_}" y name="{id_}".']
    if control == 'casilla':
        lineas.append(f'Poné primero la casilla y a su derecha un <label for="{id_}"> con el texto \'{at["nombre"]}\'.')
    else:
        lineas.append(f'Arriba del control, un <label for="{id_}"> con el texto \'{at["nombre"]}\'.')
    if at.get('obligatorio'):
        lineas.append('Es obligatorio' + (' marcarla' if control == 'casilla' else '') + ': agregale el atributo required.')
    if 'min' in at:
        lineas.append(f'El valor mínimo permitido es {at["min"]}: atributo min="{at["min"]}".')
    if 'max' in at:
        lineas.append(f'El valor máximo permitido es {at["max"]}: atributo max="{at["max"]}".')
    if 'placeholder' in at:
        lineas.append(f'Dentro del campo vacío se ve el texto de ejemplo \'{at["placeholder"]}\': atributo placeholder (no uses value).')
    if 'valores' in at:
        opciones = ', '.join(f"'{v}'" for v in at['valores'])
        lineas.append(f'Las opciones son {opciones}, en ese orden, cada una en un <option> cuyo value es igual a su texto.')
    if 'mensaje' in at:
        m = at['mensaje'].replace("'", "\\'")
        lineas.append(f'Si el valor no es válido, el navegador muestra el mensaje \'{at["mensaje"]}\': agregale al control '
                      f'exactamente oninvalid="this.setCustomValidity(\'{m}\')" oninput="this.setCustomValidity(\'\')".')
    if 'ayuda' in at:
        lineas.append(f'Debajo del control, un <small class="ayuda"> con el texto \'{at["ayuda"]}\'.')
    lineas.append('No agregues ningún otro elemento ni atributo.')
    return '\n'.join(lineas)


def verificar_campo(html, campo, id_):
    control, at = campo['control'], campo['atributos']
    sopa = BeautifulSoup(html, 'html.parser')
    p = []
    div = sopa.find('div', class_='campo')
    if div is None:
        return ['falta el <div class="campo"> que envuelve el campo']
    elemento = div.find(id=id_)
    tag_esperado = {'area': 'textarea', 'opciones': 'select'}.get(control, 'input')
    if elemento is None or elemento.name != tag_esperado:
        return [f'falta el elemento <{tag_esperado}> con id="{id_}"']
    if tag_esperado == 'input':
        tipo = CONTROLES[control][1].split('"')[1]
        if elemento.get('type') != tipo:
            p.append(f'el <input> debe tener type="{tipo}"')
    permitidos = {'id', 'name', 'type', 'required', 'min', 'max', 'placeholder', 'oninvalid', 'oninput'}
    p += atributos_de_mas(elemento, permitidos)
    if elemento.get('name') != id_:
        p.append(f'el control debe tener name="{id_}"')
    label = div.find('label', attrs={'for': id_})
    if label is None or label.get_text(strip=True) != at['nombre']:
        p.append(f'falta un <label for="{id_}"> con el texto exacto \'{at["nombre"]}\'')
    if bool(at.get('obligatorio')) != elemento.has_attr('required'):
        p.append('el control debe tener el atributo required' if at.get('obligatorio') else 'el control NO debe tener required')
    for clave in ('min', 'max'):
        if clave in at and elemento.get(clave) != str(at[clave]):
            p.append(f'el control debe tener {clave}="{at[clave]}"')
    if 'placeholder' in at and elemento.get('placeholder') != at['placeholder']:
        p.append(f'el control debe tener placeholder="{at["placeholder"]}"')
    if elemento.has_attr('value') and control not in ('casilla',):
        p.append('el control no debe tener el atributo value')
    if 'valores' in at:
        textos = [o.get_text(strip=True) for o in elemento.find_all('option')]
        if textos != at['valores']:
            p.append(f'el <select> debe tener exactamente estas opciones, en orden: {at["valores"]}')
    if 'mensaje' in at:
        oninvalid, oninput = elemento.get('oninvalid', ''), elemento.get('oninput', '')
        if 'setCustomValidity' not in oninvalid or at['mensaje'] not in oninvalid.replace("\\'", "'"):
            p.append(f'el control debe tener oninvalid="this.setCustomValidity(...)" con el mensaje \'{at["mensaje"]}\'')
        if oninput.replace(' ', '') != "this.setCustomValidity('')":
            p.append('el control debe tener exactamente oninput="this.setCustomValidity(\'\')"')
    if 'ayuda' in at:
        small = div.find('small', class_='ayuda')
        if small is None or small.get_text(strip=True) != at['ayuda']:
            p.append(f'falta un <small class="ayuda"> con el texto \'{at["ayuda"]}\'')
    return p


def generar_campo(campo, id_):
    return pedir(pedido_campo(campo, id_), lambda h: verificar_campo(h, campo, id_))


# ---------------------------------------------------------------- contenedores

def verificar_contenedor(html, tag, clase=None, textos=()):
    sopa = BeautifulSoup(html, 'html.parser')
    raiz = sopa.find(tag, class_=clase) if clase else sopa.find(tag)
    p = []
    if raiz is None:
        return [f'falta el elemento <{tag}{f" class={chr(34)}{clase}{chr(34)}" if clase else ""}>']
    for sub, texto in textos:
        e = raiz.find(sub)
        if e is None or e.get_text(strip=True) != texto:
            p.append(f'falta un <{sub}> con el texto exacto \'{texto}\'')
    if html.count(MARCA) != 1:
        p.append(f'el texto {MARCA} tiene que aparecer exactamente una vez, sin modificar')
    return p


def generar_seccion(titulo):
    pedido = (f"Generá una sección de formulario con el título '{titulo}': un <fieldset> que tenga adentro, "
              f"primero, un <legend> con el texto '{titulo}' y, después del legend, exactamente el texto {MARCA} sin modificarlo.")
    return pedir(pedido, lambda h: verificar_contenedor(h, 'fieldset', textos=[('legend', titulo)]))


def generar_fila():
    pedido = (f'Generá un contenedor que ubica campos de formulario uno al lado del otro: un <div class="fila"> '
              f'que tenga adentro solamente el texto {MARCA} sin modificarlo.')
    return pedir(pedido, lambda h: verificar_contenedor(h, 'div', clase='fila'))


def generar_boton(texto):
    pedido = f"Generá el botón que envía el formulario: un <button type=\"submit\"> con el texto '{texto}'."
    def verificar(h):
        b = BeautifulSoup(h, 'html.parser').find('button')
        if b is None or b.get('type') != 'submit' or b.get_text(strip=True) != texto:
            return [f"falta un <button type=\"submit\"> con el texto exacto '{texto}'"]
        return atributos_de_mas(b, {'type'})
    return pedir(pedido, verificar)


def generar_formulario(titulo, metodo, ruta, tema):
    pedido = (f"Generá un formulario web titulado '{titulo}' que se envía con el método {metodo.upper()} a la ruta '{ruta}': "
              f'un <form class="card tema-{tema}" method="{metodo}" action="{ruta}"> que tenga adentro, primero, '
              f"un <h2> con el texto '{titulo}' y, después del h2, exactamente el texto {MARCA} sin modificarlo.")
    def verificar(h):
        p = verificar_contenedor(h, 'form', textos=[('h2', titulo)])
        f = BeautifulSoup(h, 'html.parser').find('form')
        if f is not None:
            if f.get('method', '').lower() != metodo:
                p.append(f'el <form> debe tener method="{metodo}"')
            if f.get('action') != ruta:
                p.append(f'el <form> debe tener action="{ruta}"')
            if set(f.get('class', [])) != {'card', f'tema-{tema}'}:
                p.append(f'el <form> debe tener class="card tema-{tema}"')
            p += atributos_de_mas(f, {'class', 'method', 'action'})
        return p
    return pedir(pedido, verificar)


# ---------------------------------------------------------------- estilos (colores del tema)
# La estructura de la card (espaciado, anchos, filas) es fija y la escribe el traductor.
# El LLM interpreta la descripción del tema y genera solo los colores.

TEMAS = {
    'claro':  'fondo de la página gris muy claro; la tarjeta blanca con texto gris oscuro; '
              'campos con fondo blanco y borde gris claro; el botón gris oscuro con texto blanco',
    'oscuro': 'fondo de la página casi negro; la tarjeta gris oscuro con texto claro; '
              'campos con fondo gris un poco más claro que la tarjeta, texto claro y borde gris; el botón gris claro con texto oscuro',
    'azul':   'fondo de la página celeste muy claro; la tarjeta blanca con texto gris oscuro; '
              'el título y los legend en azul; campos con fondo blanco y borde azul claro; el botón azul con texto blanco',
}
SELECTORES_CSS = ['body', '.card', '.card h2', 'fieldset', 'legend', '.campo input', '.campo select',
                  '.campo textarea', '.ayuda', 'button']
PROPIEDADES_COLOR = {'color', 'background', 'background-color', 'border-color'}


def generar_css(tema):
    pedido = (f'Generá solo el CSS (sin etiqueta <style>) con los COLORES de un formulario con tema {tema.upper()}: '
              f'{TEMAS[tema]}.\n'
              f'Escribí reglas que cubran cada uno de estos selectores: {", ".join(SELECTORES_CSS)}.\n'
              f'Usá únicamente las propiedades {", ".join(sorted(PROPIEDADES_COLOR))}, con colores en formato #rrggbb. '
              'En .card h2 y legend usá solo la propiedad color (sin fondo). '
              'No agregues tamaños, márgenes, fuentes ni otras propiedades.')

    def verificar(css):
        p = []
        if '<' in css:
            p.append('devolvé solo CSS, sin etiquetas HTML')
        if css.count('{') != css.count('}') or css.count('{') == 0:
            p.append('las llaves { } no están balanceadas')
        reglas = re.findall(r'([^{}]+)\{([^{}]*)\}', css)
        selectores = {' '.join(s.split()) for sel, _ in reglas for s in sel.split(',')}
        faltan = [s for s in SELECTORES_CSS if s not in selectores]
        if faltan:
            p.append('faltan reglas para estos selectores: ' + ', '.join(faltan))
        otras = sorted({d.split(':')[0].strip() for _, decl in reglas for d in decl.split(';') if ':' in d}
                       - PROPIEDADES_COLOR)
        if otras:
            p.append('solo se permiten propiedades de color; sacá: ' + ', '.join(otras))
        for sel, decl in reglas:
            if {' '.join(x.split()) for x in sel.split(',')} & {'.card h2', 'legend'} and 'background' in decl:
                p.append('en .card h2 y legend no pongas fondo (background), solo color')
                break
        return p
    return pedir(pedido, verificar, es_css=True)
````

````python
%%file planilla/traductor_planilla.py
"""Traductor planilla: programa fuente -> documento HTML formateado.

1. Análisis (lexer + parser + controles semánticos): arma el árbol o informa errores.
2. Generación: recorre el árbol y le pide al agente (LLM) el HTML de cada parte y el CSS del tema.
3. Armado y formateo: une las partes y escribe el HTML con sangría prolija.
"""
import html
import re
import unicodedata
from bs4 import BeautifulSoup, NavigableString

import agente_planilla as agente
from parser_planilla import analizar, campos_de


# ---------------------------------------------------------------- identificadores de los campos

def a_identificador(texto):
    """'Fecha de nacimiento' -> 'fecha_de_nacimiento' (sin tildes ni símbolos)."""
    sin_tildes = unicodedata.normalize('NFKD', texto).encode('ascii', 'ignore').decode()
    return re.sub(r'[^a-z0-9]+', '_', sin_tildes.lower()).strip('_') or 'campo'


def asignar_ids(arbol):
    ids, usados = {}, set()
    for campo in campos_de(arbol['cuerpo']):
        base = a_identificador(campo['atributos']['nombre'])
        candidato, n = base, 2
        while candidato in usados:
            candidato, n = f'{base}_{n}', n + 1
        usados.add(candidato)
        ids[id(campo)] = candidato
    return ids


# ---------------------------------------------------------------- generación (recorre el árbol)

def generar_bloque(bloque, ids, avisar):
    if bloque['tipo'] == 'campo':
        avisar(f"  campo {bloque['control']} '{bloque['atributos']['nombre']}'")
        return agente.generar_campo(bloque, ids[id(bloque)])
    if bloque['tipo'] == 'fila':
        avisar('  fila')
        hijos = ''.join(generar_bloque(c, ids, avisar) for c in bloque['campos'])
        return agente.generar_fila().replace(agente.MARCA, hijos)
    avisar(f"  sección '{bloque['titulo']}'")
    hijos = ''.join(generar_bloque(e, ids, avisar) for e in bloque['elementos'])
    return agente.generar_seccion(bloque['titulo']).replace(agente.MARCA, hijos)


def traducir(texto, avisar=lambda mensaje: None):
    """Traduce un programa planilla a un documento HTML. Lanza ErrorPlanilla si el programa tiene errores."""
    arbol = analizar(texto)
    ids = asignar_ids(arbol)
    avisar(f"formulario '{arbol['titulo']}' (tema {arbol['tema']})")
    cuerpo = ''.join(generar_bloque(b, ids, avisar) for b in arbol['cuerpo'])
    avisar(f"  botón '{arbol['boton']}'")
    cuerpo += agente.generar_boton(arbol['boton'])
    form = agente.generar_formulario(arbol['titulo'], arbol['metodo'], arbol['ruta'], arbol['tema'])
    form = form.replace(agente.MARCA, cuerpo)
    avisar('  estilos')
    css = agente.generar_css(arbol['tema'])
    return armar_documento(arbol['titulo'], css, form)


# ---------------------------------------------------------------- armado y formateo

# Estructura fija del estilo card (la misma para todos los temas; los colores los genera el LLM).
CSS_BASE = '''
* { box-sizing: border-box; }
body { margin: 0; padding: 40px 16px; font-family: system-ui, Segoe UI, Arial, sans-serif; }
.card { max-width: 520px; margin: 0 auto; padding: 24px 28px; border-radius: 12px; box-shadow: 0 4px 18px rgba(0, 0, 0, 0.15); }
.card h2 { margin: 0 0 18px; }
fieldset { border: 1px solid; border-radius: 8px; padding: 10px 16px 2px; margin: 0 0 18px; }
legend { padding: 0 6px; font-weight: 600; }
.campo { display: flex; flex-direction: column; gap: 6px; margin-bottom: 14px; }
.campo input, .campo select, .campo textarea { width: 100%; padding: 8px 10px; border: 1px solid; border-radius: 6px; font: inherit; }
.campo textarea { min-height: 80px; resize: vertical; }
.campo:has(input[type="checkbox"]) { flex-direction: row; align-items: center; }
.campo input[type="checkbox"] { width: auto; margin: 0; }
.fila { display: flex; gap: 16px; }
.fila .campo { flex: 1; min-width: 0; }
.ayuda { font-size: 0.85em; opacity: 0.8; }
button { padding: 10px 22px; border: none; border-radius: 6px; font: inherit; font-weight: 600; cursor: pointer; }
'''

EN_UNA_LINEA = {'label', 'legend', 'option', 'button', 'h2', 'small', 'title'}
SIN_CIERRE = {'input', 'meta', 'br'}
SANGRIA = '  '


def _atributos(tag):
    partes = []
    for clave, valor in tag.attrs.items():
        if isinstance(valor, list):
            valor = ' '.join(valor)
        valor = valor.replace('&', '&amp;').replace('"', '&quot;').replace('<', '&lt;')
        partes.append(clave if valor == '' else f'{clave}="{valor}"')
    return (' ' + ' '.join(partes)) if partes else ''


def formatear_html(nodo, nivel=0):
    """Escribe cada elemento en su propia línea, con 2 espacios de sangría por nivel."""
    lineas, pre = [], SANGRIA * nivel
    for hijo in nodo.children:
        if isinstance(hijo, NavigableString):
            if hijo.strip():
                lineas.append(pre + html.escape(hijo.strip(), quote=False))
            continue
        apertura = f'<{hijo.name}{_atributos(hijo)}>'
        if hijo.name in SIN_CIERRE:
            lineas.append(pre + apertura)
        elif hijo.name in EN_UNA_LINEA or not any(not isinstance(c, NavigableString) for c in hijo.children):
            lineas.append(f'{pre}{apertura}{html.escape(hijo.get_text(" ", strip=True), quote=False)}</{hijo.name}>')
        else:
            lineas.append(pre + apertura)
            lineas.extend(formatear_html(hijo, nivel + 1))
            lineas.append(f'{pre}</{hijo.name}>')
    return lineas


def formatear_css(css, nivel):
    css = re.sub(r'/\*.*?\*/', '', css, flags=re.S)
    pre, lineas = SANGRIA * nivel, []
    for selector, declaraciones in re.findall(r'([^{}]+)\{([^{}]*)\}', css):
        lineas.append(f"{pre}{' '.join(selector.split())} {{")
        for d in declaraciones.split(';'):
            if d.strip():
                lineas.append(f"{pre}{SANGRIA}{' '.join(d.split())};")
        lineas.append(f'{pre}}}')
        lineas.append('')
    return lineas[:-1]


def armar_documento(titulo, css, form):
    lineas = ['<!DOCTYPE html>', '<html lang="es">', '  <head>', '    <meta charset="utf-8">',
              f'    <title>{html.escape(titulo)}</title>', '    <style>']
    lineas += formatear_css(CSS_BASE, 3) + [''] + ['      /* colores del tema (generados por el LLM) */'] + formatear_css(css, 3)
    lineas += ['    </style>', '  </head>', '  <body>']
    lineas += formatear_html(BeautifulSoup(form, 'html.parser'), 2)
    lineas += ['  </body>', '</html>']
    return '\n'.join(lineas) + '\n'
````

````python
%%file planilla/main.py
"""Uso: python main.py <archivo.planilla>
Traduce el programa y escribe <archivo.planilla>.html. Si hay errores, los muestra y no genera el archivo."""
import sys
from pathlib import Path

sys.path.insert(0, str(Path(__file__).parent))
from parser_planilla import ErrorPlanilla
from agente_planilla import ErrorAgente
from traductor_planilla import traducir

if hasattr(sys.stdout, 'reconfigure'):
    sys.stdout.reconfigure(encoding='utf-8')  # tildes y símbolos en la consola de Windows

if len(sys.argv) != 2:
    print(__doc__)
    sys.exit(2)

fuente = Path(sys.argv[1])
try:
    resultado = traducir(fuente.read_text(encoding='utf-8'), avisar=print)
except ErrorPlanilla as e:
    print(e)
    sys.exit(1)
except ErrorAgente as e:
    print(f'Error de generación: {e}')
    sys.exit(1)
except ConnectionError:
    print('No se pudo conectar con ollama: abrilo (y descargá el modelo con "ollama pull llama3.2") y volvé a intentar.')
    sys.exit(1)

destino = fuente.with_name(fuente.name + '.html')
destino.write_text(resultado, encoding='utf-8')
print(f'Generado: {destino}')
````

| Programas válidos | 6 | Los 5 ejemplos se traducen con el LLM y se verifica la **estructura** del HTML (método y ruta, tema, secciones, filas, tipos de control, `required`, `min`/`max`, opciones, ayuda, mensaje propio) y que el HTML salga formateado. No se compara el texto exacto porque la redacción del LLM puede variar. |
| Errores léxicos | 6 | Sangría impar, tabulación, cadena sin cerrar, palabra desconocida, carácter no reconocido, sangría que sube dos niveles |
| Errores sintácticos | 6 | Falta el botón, sección dentro de sección, atributo sin valor, botón dentro de una sección, programa sin encabezado, tema inexistente |
| Errores semánticos | 10 | Los 9 controles de la grilla semántica + varios errores informados juntos |
| Consola (`main.py`) | 2 | Caso feliz (genera el `.html`) y caso con error (mensaje con la línea, no genera archivo) |

````python
%%file planilla/test_planilla.py
"""TestSuite de planilla (pytest).

- Programas válidos: se traducen con el LLM y se verifica la ESTRUCTURA del HTML generado
  (elementos y atributos), no el texto exacto, porque la redacción del LLM puede variar.
- Errores léxicos, sintácticos y semánticos: se verifica el tipo de error y el mensaje.
- Programa completo por consola (main.py), como lo usaría un programador.
Ejecutar desde esta carpeta:  python -m pytest -v test_planilla.py
"""
import subprocess
import sys
from pathlib import Path

import pytest
from bs4 import BeautifulSoup

CARPETA = Path(__file__).parent
sys.path.insert(0, str(CARPETA))
from parser_planilla import analizar, ErrorPlanilla
from traductor_planilla import traducir

EJEMPLOS = CARPETA / 'ejemplos'


def leer(nombre):
    return (EJEMPLOS / nombre).read_text(encoding='utf-8')


def html_de(programa):
    return BeautifulSoup(traducir(programa), 'html.parser')


# ================================================================ programas válidos

def test_minimo():
    s = html_de(leer('01_minimo.planilla'))
    form = s.find('form')
    assert form['method'] == 'post' and form['action'] == 'api/contacto'
    assert 'tema-claro' in form['class']
    assert form.find('h2').get_text(strip=True) == 'Contacto'
    entrada = s.find('input', id='comentario')
    assert entrada['type'] == 'text' and entrada['placeholder'] == 'Escribí tu comentario'
    assert s.find('label', attrs={'for': 'comentario'}).get_text(strip=True) == 'Comentario'
    assert s.find('button', type='submit').get_text(strip=True) == 'Enviar'


def test_registro_secciones_y_validaciones():
    s = html_de(leer('02_registro.planilla'))
    assert 'tema-oscuro' in s.find('form')['class']
    assert [l.get_text(strip=True) for l in s.find_all('legend')] == ['Datos personales', 'Cuenta']
    edad = s.find('input', id='edad')
    assert edad['type'] == 'number' and edad['min'] == '18' and edad['max'] == '99'
    assert 'Tenés que ser mayor de edad' in edad['oninvalid']
    assert s.find('input', id='nombre').has_attr('required')
    assert s.find('input', id='correo')['type'] == 'email'
    clave = s.find('input', id='contrasena')
    assert clave['type'] == 'password'
    assert clave.find_parent('div', class_='campo').find('small', class_='ayuda').get_text(strip=True) == 'Mínimo 8 caracteres'
    casilla = s.find('input', id='acepto_los_terminos')
    assert casilla['type'] == 'checkbox' and casilla.has_attr('required')


def test_turno_filas_y_opciones():
    s = html_de(leer('03_turno.planilla'))
    assert 'tema-azul' in s.find('form')['class']
    filas = s.find_all('div', class_='fila')
    assert len(filas) == 2
    assert [i['id'] for i in filas[0].find_all('input')] == ['nombre', 'apellido']
    opciones = [o.get_text(strip=True) for o in s.find('select', id='especialidad').find_all('option')]
    assert opciones == ['Clínica', 'Pediatría', 'Traumatología']
    assert s.find('textarea', id='motivo')['placeholder'] == 'Contanos brevemente'
    assert s.find('input', id='dia_preferido')['type'] == 'date'


def test_busqueda_get_sin_secciones():
    s = html_de(leer('04_busqueda.planilla'))
    form = s.find('form')
    assert form['method'] == 'get' and form['action'] == '/buscar'
    assert s.find('fieldset') is None
    assert len(s.find_all('div', class_='fila')) == 2
    assert s.find('input', id='precio_minimo')['min'] == '0'


def test_encuesta_campos_sueltos_y_secciones():
    s = html_de(leer('05_encuesta.planilla'))
    form = s.find('form')
    hijos = [h for h in form.find_all(recursive=False)]
    # orden: título, campo suelto, sección, campo suelto, botón
    assert [h.name for h in hijos] == ['h2', 'div', 'fieldset', 'div', 'button']
    assert s.find('input', id='quiero_recibir_novedades')['type'] == 'checkbox'


def test_html_formateado_y_estilos():
    html = traducir(leer('01_minimo.planilla'))
    lineas = html.splitlines()
    assert lineas[0] == '<!DOCTYPE html>'
    assert '      <h2>Contacto</h2>' in lineas            # un elemento por línea, con sangría
    assert all((len(l) - len(l.lstrip(' '))) % 2 == 0 for l in lineas if l.strip())
    assert '.card {' in html and 'colores del tema' in html


# ================================================================ errores léxicos

E = "formulario 'C' post='x'\n"


@pytest.mark.parametrize('programa, mensaje', [
    (E + "   texto nombre='A'\n  boton 'E'\n", 'Sangría de 3 espacios'),
    (E + "\ttexto nombre='A'\n  boton 'E'\n", 'No se permiten tabulaciones'),
    ("formulario 'C post='x'\n  boton 'E'\n", 'Cadena sin cerrar'),
    (E + "  sección 'Datos'\n    texto nombre='A'\n  boton 'E'\n", "Palabra desconocida: 'sección'"),
    (E + "  texto nombre='A' @\n  boton 'E'\n", "Carácter no reconocido: '@'"),
    (E + "    texto nombre='A'\n  boton 'E'\n", 'sube más de un nivel'),
], ids=['sangria_impar', 'tabulacion', 'cadena_sin_cerrar', 'palabra_desconocida', 'caracter_raro', 'salto_de_nivel'])
def test_errores_lexicos(programa, mensaje):
    with pytest.raises(ErrorPlanilla) as e:
        analizar(programa)
    assert e.value.tipo == 'léxico'
    assert any(mensaje in m for _, m in e.value.errores)


# ================================================================ errores sintácticos

@pytest.mark.parametrize('programa, linea', [
    (E + "  texto nombre='A'\n", 3),
    (E + "  seccion 'A'\n    seccion 'B'\n      texto nombre='X'\n  boton 'E'\n", 3),
    (E + "  texto nombre=\n  boton 'E'\n", 2),
    (E + "  seccion 'A'\n    texto nombre='X'\n    boton 'E'\n", 4),
    ("texto nombre='A'\n", 1),
    ("formulario 'C' post='x' tema=rojo\n  boton 'E'\n", 1),
], ids=['falta_boton', 'seccion_en_seccion', 'atributo_sin_valor', 'boton_en_seccion', 'sin_encabezado', 'tema_inexistente'])
def test_errores_sintacticos(programa, linea):
    with pytest.raises(ErrorPlanilla) as e:
        analizar(programa)
    assert e.value.tipo in ('sintáctico', 'léxico')
    if e.value.tipo == 'sintáctico':
        assert e.value.errores[0][0] == linea


# ================================================================ errores semánticos

@pytest.mark.parametrize('cuerpo, mensaje', [
    ("  texto placeholder='hola'\n", 'no tiene nombre'),
    ("  texto nombre='A' min=3\n", "solo valen en campos 'numero'"),
    ("  numero nombre='A' min=10 max=5\n", 'es mayor que'),
    ("  opciones nombre='A'\n", 'necesita su lista de valores'),
    ("  texto nombre='A' ('x', 'y')\n", "solo vale en campos 'opciones'"),
    ("  fecha nombre='A' placeholder='x'\n", "'placeholder' no vale"),
    ("  fila\n    texto nombre='A'\n", 'al menos 2 campos'),
    ("  texto nombre='A'\n  email nombre='A'\n", 'ya se usó en la línea 2'),
    ("  texto nombre='A' obligatorio obligatorio\n", 'está repetido'),
], ids=['sin_nombre', 'min_en_texto', 'min_mayor_que_max', 'opciones_sin_lista', 'lista_en_texto',
        'placeholder_en_fecha', 'fila_de_un_campo', 'nombre_repetido', 'atributo_repetido'])
def test_errores_semanticos(cuerpo, mensaje):
    with pytest.raises(ErrorPlanilla) as e:
        analizar(E + cuerpo + "  boton 'E'\n")
    assert e.value.tipo == 'semántico'
    assert any(mensaje in m for _, m in e.value.errores)


def test_varios_errores_semanticos_se_informan_juntos():
    with pytest.raises(ErrorPlanilla) as e:
        analizar(E + "  texto placeholder='x'\n  numero nombre='N' min=9 max=1\n  boton 'E'\n")
    assert [l for l, _ in e.value.errores] == [2, 3]


# ================================================================ programa por consola (main.py)

def ejecutar_main(tmp_path, programa):
    fuente = tmp_path / 'prueba.planilla'
    fuente.write_text(programa, encoding='utf-8')
    r = subprocess.run([sys.executable, str(CARPETA / 'main.py'), str(fuente)],
                       capture_output=True, text=True, encoding='utf-8')
    return r, tmp_path / 'prueba.planilla.html'


def test_main_caso_feliz(tmp_path):
    r, salida = ejecutar_main(tmp_path, leer('01_minimo.planilla'))
    assert r.returncode == 0
    assert salida.exists()
    assert BeautifulSoup(salida.read_text(encoding='utf-8'), 'html.parser').find('form') is not None


def test_main_caso_con_error(tmp_path):
    r, salida = ejecutar_main(tmp_path, E + "  numero nombre='A' min=10 max=5\n  boton 'E'\n")
    assert r.returncode == 1
    assert 'Error semántico' in r.stdout and 'línea 2' in r.stdout
    assert not salida.exists()
````

### **Conclusiones**:

* Se diseñó e implementó **planilla**, un lenguaje de dominio específico para describir formularios web en español, con una gramática de producciones recursivas que admite infinitos programas fuente y una estructura marcada por sangría.
* **La sangría no es un problema regular:** para saber si una línea abre o cierra un bloque hay que recordar los niveles anteriores, algo que un autómata finito no puede hacer. Por eso el lexer, además de las expresiones regulares de PLY, usa una pila de niveles que genera los tokens `INDENTA` y `DESINDENTA`; con ellos, la gramática trata la sangría como si fueran llaves.
* **Separar el análisis de la generación** fue clave: el parser arma un árbol y verifica la semántica antes de llamar al LLM. Así los errores se informan completos, con su línea, y no se hacen pedidos al modelo para programas incorrectos.
* **El LLM no es confiable por sí solo.** En las pruebas, con pedidos solo "semánticos" usaba `<input>` en lugar de `<textarea>`, confundía `placeholder` con `value`, omitía etiquetas de cierre, agregaba atributos no pedidos y generaba CSS irregular. Se resolvió:
  * describiendo cada construcción con su significado **y** el elemento que le corresponde según la grilla semántica;
  * **verificando** cada respuesta y devolviéndole al LLM la lista de problemas para que la corrija;
  * usando `temperature=0` para que la salida sea reproducible (y los tests, estables);
  * dejando fija la estructura del estilo card y delegando al LLM solo los colores del tema.
* El HTML se formatea con código propio (no con el LLM), de modo que el resultado siempre es legible.
* **Limitaciones:** el traductor depende de que ollama esté instalado y abierto; la primera traducción tarda más porque se carga el modelo; y el resultado visual de los temas depende del criterio del LLM.
* **Posibles extensiones:** campos condicionales (mostrar un campo según lo elegido en otro), validaciones entre campos (confirmar contraseña) y formularios de varios pasos.
